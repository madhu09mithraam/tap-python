```python
#task 4
import re


def to_snake_case(column_name):
    name = re.sub(r'([a-z0-9])([A-Z])', r'\1_\2', column_name)
    name = re.sub(r'[^a-zA-Z0-9]+', '_', name)
    return name.strip('_').lower()
print(to_snake_case("Order ID"))
print(to_snake_case("CustomerName"))
print(to_snake_case("Freight Value"))
print(to_snake_case("OrderDate"))
print(to_snake_case("Product Category"))
```
