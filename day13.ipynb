```python
#task 1:
import os
import pandas as pd
import re
schemas = {

    "customers.csv": [
        "customer_id",
        "customer_unique_id",
        "customer_zip_code_prefix",
        "customer_city",
        "customer_state"
    ],

    "orders.csv": [
        "order_id",
        "customer_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],

    "products.csv": [
        "product_id",
        "product_category_name",
        "product_name_lenght",
        "product_description_lenght",
        "product_photos_qty",
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm"
    ],

    "sellers.csv": [
        "seller_id",
        "seller_zip_code_prefix",
        "seller_city",
        "seller_state"
    ],

    "order_items.csv": [
        "order_id",
        "order_item_id",
        "product_id",
        "seller_id",
        "shipping_limit_date",
        "price",
        "freight_value"
    ],

    "order_payments.csv": [
        "order_id",
        "payment_sequential",
        "payment_type",
        "payment_installments",
        "payment_value"
    ],

    "order_reviews.csv": [
        "review_id",
        "order_id",
        "review_score",
        "review_comment_title",
        "review_comment_message",
        "review_creation_date",
        "review_answer_timestamp"
    ],

    "geolocation.csv": [
        "geolocation_zip_code_prefix",
        "geolocation_lat",
        "geolocation_lng",
        "geolocation_city",
        "geolocation_state"
    ],

    "product_category_name_translation.csv": [
        "product_category_name",
        "product_category_name_english"
    ]
}

for file_name, required_columns in schemas.items():

    print(f"\nFile Name: {file_name}")

    print("Required Columns:")

    for column in required_columns:

        print(column)

print("\n========== SCHEMA SUMMARY ==========")

print("Total Files:", len(schemas))

for file_name, required_columns in schemas.items():

    print(f"{file_name}: {len(required_columns)} columns")

print("====================================")
```

```python
#task 2:
errors = []
for file_name in schemas:
    file_path = os.path.join(".", file_name)
    df = pd.read_csv(file_path)
    for index, row in df.iterrows():
        for column in schemas[file_name]:
            if pd.isna(row[column]):
                errors.append({
                    "file": file_name,
                    "row": index + 1,
                    "error": column + " is missing"
                })
for error in errors:
    print(error)
print("Total Errors:", len(errors))
```

```python
#task 3:
bad_records = []
for error in errors:
    file_name = error["file"]
    row_number = error["row"]
    # Read the original file
    df = pd.read_csv(file_name)
    # Get the rejected row
    row = df.iloc[row_number - 1].to_dict()
    # Add rejection details
    row["file"] = file_name
    row["row_number"] = row_number
    row["reason"] = error["error"]
    bad_records.append(row)
# Convert into DataFrame
bad_df = pd.DataFrame(bad_records)
# Save as CSV
bad_df.to_csv("bad_records.csv", index=False)
print("Rejected records saved successfully!")
```

```python
#task 4:
import pandas as pd
# Read order_items.csv
df = pd.read_csv("order_items.csv")
# Check whether price is numeric
for index, row in df.iterrows():
    price = pd.to_numeric(row["price"], errors="coerce")
    if pd.isna(price):
        errors.append({
            "file": "order_items.csv",
            "row": index + 1,
            "error": "Invalid price: " + str(row["price"])
        })
# Read orders.csv
df = pd.read_csv("orders.csv")
# Check whether order_purchase_timestamp is a valid date
for index, row in df.iterrows():
    date = pd.to_datetime(
        row["order_purchase_timestamp"],
        errors="coerce"
    )
    if pd.isna(date):
        errors.append({
            "file": "orders.csv",
            "row": index + 1,
            "error": "Invalid date: " + str(row["order_purchase_timestamp"])
        })
# Display all errors
for error in errors:
    print(error)
print("\nTotal Errors:", len(errors))
```

```python
#task 5:
df = pd.read_csv("order_items.csv")
# Validate price
for index, row in df.iterrows():
    price = pd.to_numeric(row["price"], errors="coerce")
    if pd.isna(price):
        continue
    if price <= 0:
        errors.append({
            "file": "order_items.csv",
            "row": index + 1,
            "error": "Price must be greater than 0"
        })
# Read orders.csv
df = pd.read_csv("orders.csv")
# Allowed order statuses
allowed_status = [
    "created",
    "approved",
    "invoiced",
    "processing",
    "shipped",
    "delivered",
    "canceled",
    "unavailable"
]
# Validate order status
for index, row in df.iterrows():
    if row["order_status"] not in allowed_status:
        errors.append({
            "file": "orders.csv",
            "row": index + 1,
            "error": "Invalid order status: " + str(row["order_status"])
        })
# Display all errors
for error in errors:
    print(error)
print("\nTotal Errors:", len(errors))
```

```python
#task 6:
# Regex patterns
zip_pattern = r"^\d{5}$"
id_pattern = r"^[a-fA-F0-9]{32}$"
# Validate customers.csv
df = pd.read_csv("customers.csv")
for index, row in df.iterrows():
    if not re.fullmatch(zip_pattern, str(row["customer_zip_code_prefix"])):
        errors.append({
            "file": "customers.csv",
            "row": index + 1,
            "error": "Invalid customer ZIP code prefix"
        })
    if not re.fullmatch(id_pattern, str(row["customer_id"])):
        errors.append({
            "file": "customers.csv",
            "row": index + 1,
            "error": "Invalid customer ID"
        })
# Validate sellers.csv
df = pd.read_csv("sellers.csv")
for index, row in df.iterrows():
    if not re.fullmatch(zip_pattern, str(row["seller_zip_code_prefix"])):
        errors.append({
            "file": "sellers.csv",
            "row": index + 1,
            "error": "Invalid seller ZIP code prefix"
        })
    if not re.fullmatch(id_pattern, str(row["seller_id"])):
        errors.append({
            "file": "sellers.csv",
            "row": index + 1,
            "error": "Invalid seller ID"
        })
# Validate geolocation.csv
df = pd.read_csv("geolocation.csv")
for index, row in df.iterrows():
    if not re.fullmatch(zip_pattern, str(row["geolocation_zip_code_prefix"])):
        errors.append({
            "file": "geolocation.csv",
            "row": index + 1,
            "error": "Invalid geolocation ZIP code prefix"
        })


# Validate products.csv
df = pd.read_csv("products.csv")

for index, row in df.iterrows():

    if not re.fullmatch(id_pattern, str(row["product_id"])):
        errors.append({
            "file": "products.csv",
            "row": index + 1,
            "error": "Invalid product ID"
        })


# Validate orders.csv
df = pd.read_csv("orders.csv")

for index, row in df.iterrows():

    if not re.fullmatch(id_pattern, str(row["order_id"])):
        errors.append({
            "file": "orders.csv",
            "row": index + 1,
            "error": "Invalid order ID"
        })


# Display errors
for error in errors:
    print(error)

print("\nTotal Errors:", len(errors))
```

```python
#task 7:
class SchemaError(Exception):
    pass
class FileFormatError(Exception):
    pass
for file_name, required_columns in schemas.items():
    try:
        df = pd.read_csv(file_name)
        for column in required_columns:
            if column not in df.columns:
                raise SchemaError(f"Missing required column '{column}' in {file_name}")
        print(file_name,'Schema validation done successfully.')
    except pd.errors.ParserError:
        raise FileFormatError(f"{file_name}: Invalid file format")
    except SchemaError as e:
        print(e)
    except FileNotFoundError:
        print(file_name, "not found.")
    
```

```python
#task 8:
for file_name in schemas:
    try:
        # Read CSV file
        df = pd.read_csv(file_name)
        # Check whether file is empty
        if df.empty:
            print(file_name, "is empty")
            continue
        print(file_name, "loaded successfully")
    except FileNotFoundError:
        print(file_name, "does not exist")
    except pd.errors.EmptyDataError:
        print(file_name, "is completely empty")
    except pd.errors.ParserError:
        raise FileFormatError(
            file_name + " has an invalid CSV format"
        )
```

```python
#task 9:
total_rows = 0
# Calculate total rows in all files
for file_name in schemas:
    try:
        df = pd.read_csv(file_name)
        total_rows = total_rows + len(df)
    except (FileNotFoundError, pd.errors.EmptyDataError):
        continue
# Find unique rejected rows
rejected_rows = set()
for error in errors:
    rejected_rows.add((error["file"], error["row"]))
# Calculate valid and rejected rows
rejected_count = len(rejected_rows)
valid_count = total_rows - rejected_count
# Display summary
print("VALIDATION SUMMARY")
print("----------------------")
print("Total Rows:", total_rows)
print("Valid Rows:", valid_count)
print("Rejected Rows:", rejected_count)
print("Total Errors:", len(errors))
```

```python
#task 10:
rejection_threshold = 20
for file_name in schemas:
    try:
        df = pd.read_csv(file_name)
        total_rows = len(df)
        rejected_rows = set()
        for error in errors:
            if error["file"] == file_name:
                rejected_rows.add(error["row"])
        rejected_count = len(rejected_rows)
        if total_rows > 0:
            rejection_percentage = (rejected_count / total_rows) * 100
        else:
            rejection_percentage = 0
        if rejection_percentage > rejection_threshold:
            print(file_name, "REJECTED")
            print("Rejection Percentage:", rejection_percentage, "%")
        else:
            print(file_name, "ACCEPTED")
            print("Rejection Percentage:", rejection_percentage, "%")
    except (FileNotFoundError, pd.errors.EmptyDataError):
        print(file_name, "Cannot be validated")
```
