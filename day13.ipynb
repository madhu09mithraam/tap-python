```python
#task 1:
import os

import pandas as pd

schemas = {

    "customers.csv": [
        "customer_id",
        "customer_unique_id",
        "customer_zip_code_prefix",
        "customer_city",
        "customer_state"
    ],

    "orders.csv": [
        "order_id",
        "customer_id",
        "order_status",
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],

    "products.csv": [
        "product_id",
        "product_category_name",
        "product_name_lenght",
        "product_description_lenght",
        "product_photos_qty",
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm"
    ],

    "sellers.csv": [
        "seller_id",
        "seller_zip_code_prefix",
        "seller_city",
        "seller_state"
    ],

    "order_items.csv": [
        "order_id",
        "order_item_id",
        "product_id",
        "seller_id",
        "shipping_limit_date",
        "price",
        "freight_value"
    ],

    "order_payments.csv": [
        "order_id",
        "payment_sequential",
        "payment_type",
        "payment_installments",
        "payment_value"
    ],

    "order_reviews.csv": [
        "review_id",
        "order_id",
        "review_score",
        "review_comment_title",
        "review_comment_message",
        "review_creation_date",
        "review_answer_timestamp"
    ],

    "geolocation.csv": [
        "geolocation_zip_code_prefix",
        "geolocation_lat",
        "geolocation_lng",
        "geolocation_city",
        "geolocation_state"
    ],

    "product_category_name_translation.csv": [
        "product_category_name",
        "product_category_name_english"
    ]
}

for file_name, required_columns in schemas.items():

    print(f"\nFile Name: {file_name}")

    print("Required Columns:")

    for column in required_columns:

        print(column)

print("\n========== SCHEMA SUMMARY ==========")

print("Total Files:", len(schemas))

for file_name, required_columns in schemas.items():

    print(f"{file_name}: {len(required_columns)} columns")

print("====================================")
```

```python
#task 2:
errors = []
for file_name in schemas:
    file_path = os.path.join(".", file_name)
    df = pd.read_csv(file_path)
    for index, row in df.iterrows():
        for column in schemas[file_name]:
            if pd.isna(row[column]):
                errors.append({
                    "file": file_name,
                    "row": index + 1,
                    "error": column + " is missing"
                })
for error in errors:
    print(error)
print("Total Errors:", len(errors))
```

```python
#task 3:
bad_records = pd.DataFrame(errors)
bad_records.to_csv("bad_records.csv", index=False)
print(bad_records)
print("Bad records saved to bad_records.csv")
```
