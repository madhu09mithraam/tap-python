```python
#task 7:
import pandas as pd
import yaml
import json
import os
import time
# Start pipeline
start_time = time.time()
# Step 1: Read config
with open("config.yaml") as file:
    config = yaml.safe_load(file)
# Step 2: Get paths from config
orders_path = config["paths"]["orders"]
clean_csv = config["paths"]["clean_csv"]
clean_json = config["paths"]["clean_json"]
manifest_path = config["paths"]["manifest"]
# Step 3: Read manifest
if os.path.exists(manifest_path):
    with open(manifest_path) as file:
        manifest_data = json.load(file)
else:
    manifest_data = {"processed_files": []}
# Step 4: Check if file is already processed
if orders_path in manifest_data["processed_files"]:
    print("File already processed. Skipping:", orders_path)
else:
    print("Processing:", orders_path)
    # Step 5: Read input data
    orders = pd.read_csv(orders_path)
    # Step 6: Read bad records
    bad_records = pd.read_csv("bad_records.csv")
    # Step 7: Filter order errors
    order_errors = bad_records[
        bad_records["file"] == "orders.csv"
    ]
    rejected_rows = order_errors["row_number"].unique() - 1
    # Step 8: Create cleaned data
    cleaned_orders = orders.drop(
        index=rejected_rows,
        errors="ignore"
    )
    # Step 9: Save CSV
    cleaned_orders.to_csv(clean_csv, index=False)
    # Step 10: Save JSON
    cleaned_orders.to_json(
        clean_json,
        orient="records",
        indent=4
    )
    # Step 11: Update manifest
    manifest_data["processed_files"].append(orders_path)
    with open(manifest_path, "w") as file:
        json.dump(manifest_data, file, indent=4)
    # Step 12: Calculate summary
    total_records = len(orders)
    accepted_records = len(cleaned_orders)
    rejected_records = total_records - accepted_records
    duration = round(time.time() - start_time, 2)
    summary = {
        "file": orders_path,
        "total_records": total_records,
        "accepted_records": accepted_records,
        "rejected_records": rejected_records,
        "duration_seconds": duration
    }

    # Step 13: Save summary
    with open("run_summary.json", "w") as file:
        json.dump(summary, file, indent=4)

    print("\nPipeline completed successfully!")
    print("Total records:", total_records)
    print("Accepted records:", accepted_records)
    print("Rejected records:", rejected_records)
    print("Duration:", duration, "seconds")
```
