```python
#task 4:
# Read rejected records
bad_records = pd.read_csv("bad_records.csv")

# Filter rejected orders
order_errors = bad_records[bad_records["file"] == "orders.csv"]

# Get rejected row numbers
rejected_rows = order_errors["row_number"].unique() - 1

# Get accepted orders
cleaned_orders = orders.drop(index=rejected_rows, errors="ignore")

# Save as CSV
cleaned_orders.to_csv(clean_csv, index=False)

# Save as JSON
cleaned_orders.to_json(clean_json, orient="records", indent=4)

print("Cleaned data saved successfully!")
print("Accepted records:", len(cleaned_orders))
print("Rejected records:", len(rejected_rows))
```
